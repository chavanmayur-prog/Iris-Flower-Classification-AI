# Build a Simple AI Model Using Python
## Iris Flower Classification Using a Decision Tree

**Objective:** Build a supervised machine-learning model that predicts the species of an Iris flower from four measurements.

**Dataset:** Iris dataset, a commonly used open dataset available through scikit-learn.

**Algorithm:** Decision Tree Classifier

**Workflow:** Load data → inspect and clean data → split data → train model → evaluate model → interpret results.

## 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, ConfusionMatrixDisplay
```

## 2. Load the Dataset
The dataset contains four flower measurements and a target label representing the species.

```python
iris = load_iris(as_frame=True)

X = iris.data
y = iris.target

df = X.copy()
df["species"] = y.map(dict(enumerate(iris.target_names)))

print("First five rows:")
display(df.head())

print("Dataset shape:", df.shape)
print("\nSpecies counts:")
print(df["species"].value_counts())
```

## 3. Data Inspection and Cleaning
Check for missing values and duplicate rows. The built-in Iris dataset is already structured, but these checks demonstrate basic data-cleaning steps.

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:", df.duplicated().sum())

# Remove duplicate rows if any are present.
df = df.drop_duplicates().reset_index(drop=True)

# Separate features and target after cleaning.
X = df[iris.feature_names]
y = df["species"].map({name: i for i, name in enumerate(iris.target_names)})

print("\nShape after removing duplicates:", df.shape)
```

## 4. Split the Data
Use 80% of the data for training and 20% for testing. Stratification keeps the class proportions similar in both sets.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 5. Build and Train the Model
A Decision Tree learns rules from the training data to predict the flower species.

```python
model = DecisionTreeClassifier(max_depth=4, random_state=42)
model.fit(X_train, y_train)

print("Model training completed.")
```

## 6. Evaluate the Model
Accuracy measures the proportion of test predictions that are correct. The confusion matrix shows correct and incorrect predictions for each species.

```python
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)

print(f"Test accuracy: {accuracy:.2%}")
print("\nClassification report:")
print(classification_report(y_test, y_pred, target_names=iris.target_names))

disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=iris.target_names)
disp.plot()
plt.title("Confusion Matrix - Iris Classification")
plt.show()
```

## 7. Understand Feature Importance
Feature importance gives an indication of how much each measurement contributes to the trained Decision Tree. These values describe this model and dataset; they are not universal rules.

```python
importance = pd.Series(
    model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

display(importance.to_frame(name="Importance"))

importance.sort_values().plot(kind="barh")
plt.xlabel("Feature importance")
plt.title("Decision Tree Feature Importance")
plt.show()
```

## 8. Predict a New Example
Enter hypothetical measurements in centimetres. The model predicts one of the three Iris species.

```python
# Example measurements: sepal length, sepal width, petal length, petal width (cm)
new_flower = pd.DataFrame(
    [[5.1, 3.5, 1.4, 0.2]],
    columns=iris.feature_names
)

prediction = model.predict(new_flower)[0]
print("Predicted species:", iris.target_names[prediction])
```

## 9. Results and Interpretation

- The model learns from labelled examples, so this is **supervised learning**.
- Test accuracy is printed above and depends on the train/test split and model settings.
- The confusion matrix shows which species were classified correctly and where errors occurred.
- Feature importance indicates which measurements the trained Decision Tree relied on most.
- The example prediction demonstrates how the model can classify a new flower from its measurements.

## 10. Conclusion

A Decision Tree model was built using the Iris dataset. The workflow included loading data, checking for missing values and duplicates, splitting the dataset, training the model, evaluating its performance, and making a sample prediction. This small project demonstrates the basic machine-learning process. Its results apply to this dataset and should not be assumed to generalize to every flower dataset.
