## Iris Flower Classification Using a Decision Tree

**Objective:** Build a supervised machine-learning model to predict Iris flower species using four measurements.

**Dataset:** Iris dataset from scikit-learn  
**Algorithm:** Decision Tree Classifier  
**Tools:** Python, Pandas, Matplotlib, Scikit-learn

### Steps
1. Load and inspect data.
2. Check missing values and duplicate rows.
3. Split data into training and testing sets.
4. Train a Decision Tree model.
5. Evaluate accuracy and confusion matrix.
6. Interpret results and predict a new flower.

## 1. Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

print("All libraries imported successfully!")
```

## 2. Load the Dataset

```python
iris = load_iris(as_frame=True)
df = iris.frame.copy()
df["species"] = df["target"].map(dict(enumerate(iris.target_names)))
df = df.drop(columns=["target"])

print("First five records:")
display(df.head())
print("Dataset shape:", df.shape)
print("\nSpecies counts:")
print(df["species"].value_counts())
```

## 3. Data Cleaning

```python
print("Missing values in each column:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())

# Remove duplicate rows if any are found.
df = df.drop_duplicates().reset_index(drop=True)

X = df[iris.feature_names]
y = df["species"].map({name: i for i, name in enumerate(iris.target_names)})

print("\nDataset shape after cleaning:", df.shape)
```

## 4. Split into Training and Testing Sets
Use 80% of the data for training and 20% for testing. Stratification keeps the class proportions similar.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 5. Build and Train the Model

```python
model = DecisionTreeClassifier(max_depth=4, random_state=42)
model.fit(X_train, y_train)
print("Decision Tree model trained successfully.")
```

## 6. Evaluate the Model

```python
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Model accuracy: {accuracy:.2%}")

print("\nClassification report:")
print(classification_report(y_test, y_pred, target_names=iris.target_names))

cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=iris.target_names)
disp.plot()
plt.title("Confusion Matrix - Iris Classification")
plt.show()
```

## 7. Feature Importance

```python
importance = pd.Series(model.feature_importances_, index=iris.feature_names).sort_values(ascending=False)
display(importance.to_frame(name="Importance"))

importance.sort_values().plot(kind="barh")
plt.xlabel("Importance")
plt.title("Feature Importance in the Decision Tree")
plt.show()
```

## 8. Predict a New Flower

```python
# Example measurements in centimetres:
# sepal length, sepal width, petal length, petal width
new_flower = pd.DataFrame(
    [[5.1, 3.5, 1.4, 0.2]],
    columns=iris.feature_names
)
predicted_class = model.predict(new_flower)[0]
print("Predicted Iris species:", iris.target_names[predicted_class])
```

## 9. Results and Interpretation

- The model uses supervised learning because it learns from labelled examples.
- Test accuracy shows the proportion of test samples classified correctly.
- The confusion matrix shows correct predictions and misclassifications for each species.
- Feature importance indicates which measurements the trained Decision Tree relied on most.
- The sample prediction demonstrates classification of a flower with new measurements.
